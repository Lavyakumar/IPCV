```python
img =cv2.imread('/content/letter.jpeg', 0 ) #grayscale0
```

now we will blur the img

```python
kernel= np.ones((5,5), np.float32)/25 # 5x5 ka kernal bnaya all 1/25
img_blur= cv2.filter2D(img, -1, kernel) #-1 is a flag which says that value dsnt chnge
```

```python
cv2.imwrite("1.png", np.abs(img_blur))
```

```python
noise= 20*np.sin(np.linspace(0,50,img.shape[1])) #1 column 0 means row
```

```python
noise = noise.reshape(1,-1) #chnage to 2D
```

```python
#if (2,-1 ).  then if 10 elements so 2 row 5 columns
```

```python
noisy = img_blur+noise
```

```python
cv2.imwrite("2.png", np.abs(noisy))
```

Here's how you can clamp the `noisy` image's pixel values to the valid range [0, 255] and then convert it to an 8-bit unsigned integer type (`uint8`).

```python
noisy = np.clip(noisy, 0, 255).astype(np.uint8)
```

```python
f= np.fft.fft2(noisy)#(H freq abhi center mai h) fft algo mai ff2 fn for 2D img to conver img into frequency spectrum
```

```python
fshift= np.fft.fftshift(f) # low mid mai aa gai
```

```python
h=np.fft.fft2(kernel,s=noisy.shape)
h_shift=np.fft.fftshift(h) #5x5 value is there else is 0
```

```python
epsilon= 1e-5
```

```python
h_shift=np.where(h_shift ==0,epsilon, h_shift)
```

```python
g=fshift/h_shift
```

```python
g_ishift=np.fft.ifftshift(g) #inverse
```

```python
g_iift=np.fft.ifft2(g_ishift) #inverse
```

```python
inv_img= np.abs(g_iift)
```

```python
cv2.imwrite("3.png", np.abs(inv_img))
```

### `np.fft.fft(a, n=None, axis=-1, norm=None)`
*   **Definition:** Computes the one-dimensional Discrete Fourier Transform (DFT).
*   **Notes:** Converts a signal from its original domain (e.g., time or space) to a frequency domain representation. For 2D images, use `np.fft.fft2`.

### `np.fft.fftshift(x, axes=None)`
*   **Definition:** Shifts zero-frequency component to the center of the spectrum.
*   **Notes:** Useful for visualizing the Fourier transform, as low-frequency components (which are often the most important) are moved from the corners to the center of the output.

### `np.fft.ifft(a, n=None, axis=-1, norm=None)`
*   **Definition:** Computes the one-dimensional inverse Discrete Fourier Transform (IDFT).
*   **Notes:** Converts a signal from the frequency domain back to its original domain. For 2D images, use `np.fft.ifft2`.

### `np.fft.ifftshift(x, axes=None)`
*   **Definition:** Undoes the effect of `fftshift`.
*   **Notes:** Shifts the zero-frequency component back to the original (corner) position, which is necessary before applying the inverse Fourier transform to reconstruct the original signal.
